```python
#Q11:What is list comprehension? Write a Python program to create a list of squares of numbers from 1 to 10 using list comprehension
squares = [number ** 2 for number in range(1, 11)]

print(squares)
```

```python
#Q12:Write a Python program that removes duplicates from a list [1, 2, 2, 3, 3, 4] without using the set() function.
numbers = [1, 2, 2, 3, 3, 4]

unique_numbers = []

for number in numbers:
    if number not in unique_numbers:
        unique_numbers.append(number)

print(unique_numbers)
```
